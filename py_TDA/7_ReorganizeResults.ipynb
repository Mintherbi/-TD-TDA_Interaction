```python
import os
import shutil

# Base directory
base_dir = "../Results/TDA_Iteration"

# Expected file names map to new folder names
file_mapping = {
    "1_UMAP_3D.png": "1_UMAP_3D",
    "2_Persistence_Diagram.png": "2_Persistence_Diagram",
    "3_Transition_Network.png": "3_Transition_Network",
    "4_Manifold_Skeleton.png": "4_Manifold_Skeleton",
    "5_HighDim_Persistence.png": "5_HighDim_Persistence" # In case it exists
}

# Get all items in base_dir
items = os.listdir(base_dir)

# Filter for directories (these are the regions)
regions = [d for d in items if os.path.isdir(os.path.join(base_dir, d)) and d not in file_mapping.values()]

print(f"Found regions: {regions}")

# Create new directories
for folder_name in file_mapping.values():
    folder_path = os.path.join(base_dir, folder_name)
    os.makedirs(folder_path, exist_ok=True)
    print(f"Created directory: {folder_path}")

# Move files
for region in regions:
    region_path = os.path.join(base_dir, region)
    
    # Check files in the region folder
    for filename in os.listdir(region_path):
        if filename in file_mapping:
            src = os.path.join(region_path, filename)
            
            # New destination: ../Results/TDA_Iteration/{GraphType}/{Region}.png
            target_folder = file_mapping[filename]
            dst_filename = f"{region}.png"
            dst = os.path.join(base_dir, target_folder, dst_filename)
            
            # Copy file (safer than move initially)
            shutil.copy2(src, dst)
            print(f"Copied: {src} -> {dst}")
            
print("\nReorganization complete. Please check the new folders.")
print("If satisfied, you can manually delete the old region folders.")
```
