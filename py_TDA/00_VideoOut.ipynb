```python
# --- 1. Imports & Setup ---
import os
import numpy as np
import cv2
from pathlib import Path
from allensdk.brain_observatory.ecephys.ecephys_project_cache import EcephysProjectCache

print("Libraries imported.")
```

```python
# --- 2. Load Movie Data ---
output_dir = "../Dataset/RAW"
manifest_path = os.path.join(output_dir, "manifest.json")
cache = EcephysProjectCache.from_warehouse(manifest=manifest_path)

print("Loading Natural Movie One Template...")
movie_template = cache.get_natural_movie_template(1)
print(f"Movie Template Shape: {movie_template.shape}")
# Shape is typically (Frames, Height, Width)
```

```python
# --- 3. Export to Video (MP4) ---
save_dir = "../Dataset/Processed"
os.makedirs(save_dir, exist_ok=True)
save_path = os.path.join(save_dir, "natural_movie_one.mp4")

# Video Settings
fps = 30.0
n_frames, height, width = movie_template.shape
print(f"Exporting {n_frames} frames at {fps} FPS to {save_path}...")

# Initialize VideoWriter
# 'mp4v' is a safe codec for .mp4. On macOS 'avc1' might also work well.
fourcc = cv2.VideoWriter_fourcc(*'mp4v') 
out = cv2.VideoWriter(save_path, fourcc, fps, (width, height), isColor=False)

if not out.isOpened():
    print("Error: Could not open video writer.")
else:
    for i, frame in enumerate(movie_template):
        # Normalize frame to 0-255 uint8
        # The raw template might be float or have different range, so we normalize per frame or globally.
        # Usually natural movie is 0-255 or 0-1. Let's normalize safely.
        frame_norm = cv2.normalize(frame, None, 0, 255, cv2.NORM_MINMAX).astype('uint8')
        
        out.write(frame_norm)
        
        if i % 100 == 0:
            print(f"Processed {i}/{n_frames} frames", end='\r')

    out.release()
    print(f"\nSuccessfully saved video to: {os.path.abspath(save_path)}")
```
